```python
import numpy as np

# 학습 데이터 준비
x_train = np.array([1, 2, 3], dtype=np.float32)  # 입력 (사과 개수)
y_train = np.array([1, 2, 3], dtype=np.float32)  # 출력 (가격)

# 모델의 초기 가중치 설정(weight와 bias의 초기값을 random하게 설정)
weight = np.random.rand()
bias = np.random.rand()

# 학습률 및 학습 횟수 설정
learning_rate = 0.01
epochs = 2000

# 경사하강법 학습 과정
for epoch in range(epochs):
    # 예측값 계산
    y_pred = weight * x_train + bias

    # 현 시점의 cost 값 계산(손실함수)
    cost = np.mean((y_pred - y_train) ** 2)

    # 미분을 이용한 기울기 계산
    weight_grad = np.mean(2 * (y_pred - y_train) * x_train)
    bias_grad = np.mean(2 * (y_pred - y_train))

    # weight와 bias 업데이트
    weight -= learning_rate * weight_grad
    bias -= learning_rate * bias_grad

    # 학습 과정 출력 (100번마다)
    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch + 1:4d}: Weight = {weight:.4f}, Bias = {bias:.4f}, Cost = {cost:.4f}")

# 학습 완료 후 예측 함수
def predict(apples):
    return weight * apples + bias

# 테스트: 사과 6개, 10개일 때 예측
print("\n예측 결과:")
print(f"사과 6개일 때 예상 가격: {predict(6):.4f}")
print(f"사과 10개일 때 예상 가격: {predict(10):.4f}")
```
