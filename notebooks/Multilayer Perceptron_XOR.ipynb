```python
import numpy as np
import pandas as pd
import tensorflow as tf
import random
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam

np.random.seed(0)
tf.random.set_seed(0)
random.seed(0)

# 학습 데이터 (XOR 연산)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([[0], [1], [1], [0]])

# 모델 구성
model = Sequential([
    Input(shape=(2,)),
    Dense(2, activation='sigmoid'),  # 은닉층: 2개 퍼셉트론
    Dense(1, activation='sigmoid')   # 출력층: 1개 퍼셉트론
])

# 모델 컴파일 및 학습
model.compile(loss='binary_crossentropy',
              optimizer=Adam(learning_rate=0.1),
              metrics=['accuracy'])

model.fit(X, y, epochs=300, verbose=0)

# 은닉층과 출력층 가중치 및 바이어스 가져오기
hidden_weights, hidden_bias = model.layers[0].get_weights()  # (2, 2), (2,)
output_weights, output_bias = model.layers[1].get_weights()  # (2, 1), (1,)

# 세 퍼셉트론의 매개변수 출력
print("\n[퍼셉트론 A (은닉층 1번 노드)]")
print(f"  가중치: {hidden_weights[:, 0]}")
print(f"  바이어스: {hidden_bias[0]:.4f}")

print("\n[퍼셉트론 B (은닉층 2번 노드)]")
print(f"  가중치: {hidden_weights[:, 1]}")
print(f"  바이어스: {hidden_bias[1]:.4f}")

print("\n[퍼셉트론 C (출력층 노드)]")
print(f"  가중치: {output_weights[:, 0]}")
print(f"  바이어스: {output_bias[0]:.4f}")

# 테스트 출력
print("\n[학습 완료된 XOR 연산자 모델 테스트]")
for x in X:
    prob = model.predict(np.array([x]), verbose=0)[0][0]
    pred = int(prob >= 0.5)
    print(f"입력: {x} → 예측: {pred} (확률: {prob:.4f})")



# 내부 노드 출력 및 최종 출력 확인을 위해 중간값 포함하여 출력값을 수동 계산해보는 코드
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

results = []
for x in X:
    # 은닉층 계산
    z1 = np.dot(x, hidden_weights) + hidden_bias
    a1 = sigmoid(z1)

    # 출력층 계산
    z2 = np.dot(a1, output_weights) + output_bias
    a2 = sigmoid(z2)

    pred = int((a2[0] >= 0.5))
    results.append({
        'x1': x[0], 'x2': x[1],
        'a1_1': round(a1[0], 4), 'a1_2': round(a1[1], 4),
        'a2': round(a2[0], 4),
        '예측': pred
    })

# 결과 테이블 출력
df = pd.DataFrame(results)
print("\n\n[퍼셉트론 내부 노드 출력 분석]")
print(df)
```
