```python
!pip -q install transformers==5.15.0 gradio==6.24.0 ffmpeg-python==0.2.0 accelerate==1.14.0

from transformers import pipeline
import gradio as gr
import ffmpeg

asr = pipeline("automatic-speech-recognition", model="openai/whisper-small", device_map="auto")

def transcribe(video):
    (
        ffmpeg
        .input(video)
        .audio
        .output("audio.wav", ar=16000, ac=1, t=30)
        .overwrite_output()
        .run(quiet=True)
    )
    return asr("audio.wav")["text"]

gr.Interface(
    fn=transcribe,
    inputs=gr.Video(label="🎬 비디오 업로드"),
    outputs=gr.Textbox(label="📝 자막", lines=10, placeholder="생성된 자막이 여기에 표시됩니다."),
    title="영상 자막 생성 AI"
).launch()
```
