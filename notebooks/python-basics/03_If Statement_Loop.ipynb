```python
# 1. if 조건문

bp = 135

if bp >= 130:
    print("고혈압입니다.")
else:
    print("정상 범위입니다.")
```

```python
# 2. for 반복문

# 가장 단순한 반복문
for i in range(10):
    print(i)

print("\n")

# 리스트 데이터의 요소값 순회
ages = [25, 32, 47, 51]
for age in ages:
    print(f"Age: {age}")
```

```python
# 3. 인덱스와 값 동시 사용

bp_list = [118, 130, 140, 125]

for idx, bp in enumerate(bp_list):
    print(f"{idx}번째 사람: {bp} mmHg")
```

```python
# 4. while 반복문

count = 1
while count <= 5:
    print("Count:", count)
    count += 1
```

```python
# 5. break와 continue

bp_list = [118, 130, 140, 125]

# break 예시
for bp in bp_list:
    if bp >= 140:
        print("위험: 측정 종료")
        break
    print("측정값:", bp)

# continue 예시
for bp in bp_list:
    if bp < 130:
        continue  # 130 미만은 건너뛰기
    print("High BP:", bp)
```

```python
# 응용 예시 1 — 고혈압 대상자 출력

checkups = [
    {"name": "Kim", "age": 32, "bp": 120},
    {"name": "Lee", "age": 47, "bp": 140},
    {"name": "Park", "age": 25, "bp": 118}
]

for row in checkups:
    if row["bp"] >= 130:
        print(f"{row['name']}: {row['bp']} mmHg → 고혈압")
```

```python
# 응용 예시 2 — 고혈압 & 40세 이상

for row in checkups:
    if row["bp"] >= 130 and row["age"] >= 40:
        print(f"{row['name']} → 고혈압 + 40세 이상")
```

```python
# 응용 예시 3 — 첫번째 고혈압 환자 찾기

for row in checkups:
    if row["bp"] >= 130:
        print("첫번째 발견:", row)
        break
```
