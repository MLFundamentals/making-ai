```python
# 1. Series 생성

import pandas as pd

ages = pd.Series([25, 32, 47, 51])
print(ages)
print("평균 나이:", ages.mean())
```

```python
# 2. DataFrame 생성

data = {
    "name": ["Kim", "Lee", "Park", "Choi"],
    "age": [25, 32, 47, 51],
    "bp": [118, 130, 140, 125]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 3. 행/열 조회

# 열(column) 조회
print(df["name"])          # Series 반환
print()
print(df[["name", "bp"]])  # DataFrame 반환
print()

# 행(row) 조회
print(df.loc[1])           # 라벨 기반
print()
print(df.iloc[2])          # 인덱스 기반
```

```python
# 4. 조건 필터링

# 혈압이 130 이상인 행만
high_bp = df[df["bp"] >= 130]
print(high_bp)
```

```python
# 5. 결측치 처리

df2 = pd.DataFrame({
    "age": [25, None, 47, 51],
    "bp": [118, 130, None, 125]
})

# 결측치 확인
print(df2.isnull())
print()

# 결측치 제거
print(df2.dropna())
print()

# 결측치 채우기(평균으로 대체)
df2["bp"] = df2["bp"].fillna(df2["bp"].mean())
print(df2)
```

```python
# 응용 예시 — CSV 읽기/쓰기

# CSV 저장
df.to_csv("checkup.csv", index=False, encoding="utf-8")

# CSV 읽기
df_loaded = pd.read_csv("checkup.csv")
print(df_loaded)
```

```python
# 응용 예시 — 통계 요약

print(df.describe())        # 숫자열 요약 통계
print("최대 혈압:", df["bp"].max())
print("나이 표준편차:", df["age"].std())
```
