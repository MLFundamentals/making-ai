```python
# 1. 코드 셀 실행, 출력, 주석
# 한 줄 주석: # 뒤는 설명으로 처리된다.

print(3 + 4)                # 덧셈
print(10 / 3)               # 나눗셈(부동소수점)
print(10 // 3)              # 정수 나눗셈(몫)
print(10 % 3)               # 나머지
print(2 ** 5)               # 거듭제곱
```

```python
# 2. 변수와 자료형, 형 변환

# 변수에 값 저장
age = 40               # 정수(int)
height = 1.70          # 실수(float) (단위: m)
name = "Kim"           # 문자열(str)

# 자료형 확인
print(type(age), type(height), type(name))

# 형 변환
s = "42"               # 문자열
n = int(s)             # 정수로 변환
x = float("3.14")      # 실수로 변환
print(n + 8, x * 2)

# f-string으로 포매팅
print(f"Name: {name}, Age next year: {age + 1}")
```

```python
# 3. 입력과 출력(주의: input()은 항상 문자열)

# 숫자 입력. 예) 170 (cm), 65 (kg)
h_cm = input("키(cm)를 입력하세요: ")
w_kg = input("몸무게(kg)를 입력하세요: ")

# input()은 문자열을 반환하므로 계산 전에 숫자로 변환해야함
h_m = float(h_cm) / 100
w = float(w_kg)

bmi = w / (h_m ** 2)

# 소수점 둘째자리로 반올림: round(값, 2)
print(f"Your BMI is {round(bmi, 2)}")
```

```python
# 4. 문자열 인덱싱과 슬라이싱

# 샘플 문자열
text = "사람사람이 모두 금덩어리 아님이 없는데 자기가 착각해서 그런 줄 모르고 있는 것입니다."

print(text[0])     # 첫 글자
print(text[:5])    # 처음 5글자
print(text[-1])    # 마지막 글자
print(text[-5:])   # 마지막 5글자
print("문장 길이:", len(text))
```

```python
# 5. 문자열 메서드(함수) 활용

text = "사람사람이 모두 금덩어리 아님이 없는데 자기가 착각해서 그런 줄 모르고 있는 것입니다."

print(text.count("사람"))       # 특정 단어 개수
print(text.find("금덩어리"))    # 위치 찾기
print(text.split())             # 공백 기준 분리
print(text.replace("있는 것입니다", "있습니다"))  # 치환
```

```python
# 응용 예시 - 미니 건강검진 결과 요약 출력

name = "Kim"
age = 32
sbp = 130     # 수축기 혈압 (systolic BP)
dbp = 82      # 이완기 혈압 (diastolic BP)

pulse_pressure = sbp - dbp  # 맥압
mean_arterial_pressure = dbp + (pulse_pressure / 3)

print(f"[Checkup Summary]")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"SBP/DBP: {sbp}/{dbp} mmHg")
print(f"Pulse Pressure: {pulse_pressure} mmHg")
print(f"MAP ≈ {round(mean_arterial_pressure, 1)} mmHg")
```
