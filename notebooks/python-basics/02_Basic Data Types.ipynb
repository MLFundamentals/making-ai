```python
# 자료형 1 : 리스트

# 리스트 생성
ages = [25, 32, 47, 51]

# 인덱싱 (0부터 시작)
print(ages[0])   # 첫번째 값
print(ages[-1])  # 마지막 값

# 슬라이싱
print(ages[1:3]) # index 1~2까지

# 값 수정
ages[2] = 48
print(ages)

# 추가/삭제
ages.append(60)  # 끝에 추가
ages.remove(32)  # 값으로 삭제
print(ages)
```

```python
# 자료형 2 : 튜플

# 튜플은 변경 불가능(immutable)한 리스트
sizes = (170, 65)  # (키 cm, 몸무게 kg)
print(sizes[0], "cm")
# sizes[0] = 175  # ❌ 오류 발생

# 단일 값 튜플 만들기 → 콤마 필수
one_item = (42,)
print(one_item, type(one_item))
```

```python
# 자료형 3 : 딕셔너리

person = {
    "name": "Kim",
    "age": 32,
    "bp": 120
}

print(person["name"])      # 조회
person["bp"] = 125         # 수정
person["height"] = 172     # 새로운 키-값 추가
print(person)

# 딕셔너리 메서드
print(person.keys())
print(person.values())
print(person.items())
```

```python
# 응용 예시 - 리스트 안의 딕셔너리

# 건강검진 데이터 (간단 버전)
checkups = [
    {"name": "Kim", "age": 32, "bp": 120},
    {"name": "Lee", "age": 47, "bp": 140},
    {"name": "Park", "age": 25, "bp": 118}
]

# 모든 사람의 이름과 혈압 출력
for row in checkups:
    print(f"{row['name']}: {row['bp']} mmHg")
```

```python
# 응용 예시 - 혈압이 높은 사람 찾기

high_bp = []  # 결과 저장할 리스트

for row in checkups:
    if row["bp"] >= 130:
        high_bp.append(row["name"])

print("고혈압 대상자:", high_bp)
```
