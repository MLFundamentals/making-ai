```python
# 3. 기본값 매개변수

def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # 기본값 사용
greet("Kim")     # 인자 전달
```

```python
# 응용 예시 — 합격자 점수만 모으는 함수

def get_pass_scores(scores, cutoff):
    """입력받은 리스트 scores에서 cutoff 점수 이상의 값만
    새로운 리스트에 담아 반환하는 함수"""
    passed = []  # 합격자 점수를 저장할 리스트

    for score in scores:
        if score >= cutoff:
            passed.append(score)

    return passed

exam_scores = [72, 88, 64, 91, 55, 79]
pass_scores = get_pass_scores(exam_scores, 70)

print("합격자 점수:", pass_scores)
```

```python
# 응용 예시 — 평균 계산 함수

def average(values):
    """리스트 값들의 평균 계산"""
    return sum(values) / len(values)

ages = [25, 32, 47, 51]
print("평균 나이:", average(ages))
```
