```python
# 1. 기본 선 그래프

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)               # 선 그래프
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Simple Line Plot")
plt.show()
```

```python
# 2. 산점도

age = [25, 32, 47, 51]
bp = [118, 130, 140, 125]

plt.scatter(age, bp, color="blue", marker="o")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("Age vs Blood Pressure")
plt.show()
```

```python
# 3. 히스토그램

bp_data = [118, 130, 140, 125, 135, 128, 150, 110]

plt.hist(bp_data, bins=5, color="skyblue", edgecolor="black")
plt.xlabel("Blood Pressure")
plt.ylabel("Frequency")
plt.title("Distribution of Blood Pressure")
plt.show()
```

```python
# 4. 여러 데이터 비교

ages = [25, 32, 47, 51]
bp_male = [118, 130, 140, 125]
bp_female = [115, 128, 135, 122]

plt.plot(ages, bp_male, label="Male", marker="o")
plt.plot(ages, bp_female, label="Female", marker="s")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("BP by Age and Gender")
plt.legend()
plt.show()
```

```python
# 응용 예시 — 건강검진 데이터 시각화

import pandas as pd

df = pd.DataFrame({
    "age": [25, 32, 47, 51, 38, 42],
    "bp": [120, 125, 128, 132, 138, 142]
})

# 산점도
plt.scatter(df["age"], df["bp"], color="green")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("Health Checkup: Age vs BP")
plt.show()

# 히스토그램
plt.hist(df["bp"], bins=5, color="orange", edgecolor="black")
plt.xlabel("Blood Pressure")
plt.ylabel("Frequency")
plt.title("BP Distribution")
plt.show()
```
