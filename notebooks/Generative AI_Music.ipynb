```python
!pip -q install transformers==5.15.0 accelerate==1.14.0

from transformers import pipeline
from IPython.display import Audio
import numpy as np

# 파이프라인 로드
pipe = pipeline("text-to-audio", model="facebook/musicgen-small", device_map="auto")

# 프롬프트 (자유롭게 수정해보세요!)
prompt = "catchy K-pop track with bright synths and dance beat"

# 음악 생성
out = pipe(prompt, forward_params={"do_sample": True, "max_new_tokens": 1024})

# 재생
Audio(np.array(out["audio"]).squeeze(), rate=out.get("sampling_rate", 32000))

# 프롬프트 예: "emotional rock ballad", "a relaxing piano melody with soft strings", "classical string quartet in baroque style"
```
