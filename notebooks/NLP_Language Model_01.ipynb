```python
import re
from collections import defaultdict, Counter

# 텍스트 전처리: 소문자로 변환하고 특수문자 제거
def preprocess(text):
    text = text.lower()  # 모든 문자를 소문자로 변환
    text = re.sub(r'[^\w\s]', '', text)  # 알파벳, 숫자, 공백을 제외한 특수문자 제거
    return text.split()  # 공백 기준으로 토큰 분리하여 단어 리스트 반환

# N-그램(기본값: 3-그램) 생성 함수
def generate_ngrams(words, n=3):
    ngrams = []
    for i in range(len(words) - n + 1):
        ngram = tuple(words[i:i + n])  # 연속된 n개의 단어를 하나의 튜플로 묶기
        ngrams.append(ngram)
    return ngrams

# N-그램 언어 모델 학습 함수
def train_ngram_model(text, n=3):
    words = preprocess(text)  # 텍스트 전처리 후 단어 리스트 생성
    ngrams = generate_ngrams(words, n)  # N-그램 리스트 생성

    model = defaultdict(Counter)  # prefix(앞 n-1 단어)에 대해 다음 단어 빈도를 저장할 딕셔너리

    for ngram in ngrams:
        prefix, target = ngram[:-1], ngram[-1]  # 앞의 n-1 단어를 prefix, 마지막 단어를 target으로 분리
        model[prefix][target] += 1  # prefix에 대한 target 단어의 출현 빈도 카운트

    return model  # 학습된 언어 모델 반환

# 다음 단어 예측 함수
def predict_next_word(model, prefix):
    if prefix in model:
        # prefix에 대해 가장 빈도가 높은 다음 단어 반환
        return model[prefix].most_common(1)[0][0]
    else:
        return None  # 학습 데이터에 없는 prefix인 경우 None 반환

# 예제 텍스트
text = "I love natural language processing and natural language models are powerful for understanding text."

# 3-그램 모델 학습
ngram_model = train_ngram_model(text, n=3)

# 특정 prefix에 대한 다음 단어 예측
prefix = ('natural', 'language')  # 앞의 두 단어(prefix)
next_word = predict_next_word(ngram_model, prefix)

print(f"'{prefix}' 다음에 나올 확률이 높은 단어: '{next_word}'")
```
