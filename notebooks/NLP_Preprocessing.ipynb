```python
# 1. 예제 문장 정의
sentences = [
    "나는 머신러닝을 정말 좋아한다.",
    "자연어 처리는 정말 재미있다!",
    "파이썬으로 자연어 처리를 해보자."
]

# 2. 토큰화 함수 정의
import re   # 정규표현식 모듈 호출

def tokenize(sentence):
    sentence = re.sub(r'[^가-힣0-9\s]', ' ', sentence)  # 특수문자 제거
    tokens = []
    for token in sentence.split():
        if token:  # 빈 문자열 제거
            tokens.append(token)
    return tokens

# 각 문장에 토큰화 적용
tokenized_sentences = []
for s in sentences:
    tokenized = tokenize(s)
    tokenized_sentences.append(tokenized)

print("① 토큰화된 문장:")
for tokens in tokenized_sentences:
    print(tokens)

# 3. 어휘 목록(vocabulary) 생성
vocab_set = set()
for tokens in tokenized_sentences:
    for token in tokens:
        vocab_set.add(token)
vocab = sorted(list(vocab_set))

print("\n② 어휘 목록:")
print(vocab)

# 4. 단어 → 인덱스 매핑 생성
word2idx = {}
for idx, word in enumerate(vocab):
    word2idx[word] = idx + 1  # 인덱스 번호에서 0은 별도용

print("\n③ 단어-인덱스 매핑:")
print(word2idx)

# 5. 문장을 인덱스 시퀀스로 변환
indexed_sentences = []
for tokens in tokenized_sentences:
    indexed_seq = []
    for token in tokens:
        index = word2idx.get(token, 0)  # 없으면 0
        indexed_seq.append(index)
    indexed_sentences.append(indexed_seq)

print("\n④ 인덱스 시퀀스로 변환된 문장:")
for seq in indexed_sentences:
    print(seq)
```
