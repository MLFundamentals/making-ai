```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# MNIST 데이터 로드
(X_train, y_train), _ = tf.keras.datasets.mnist.load_data()

sample_img = X_train[20000]
sample_label = y_train[20000]

# 픽셀 경계를 설정하기 위한 좌표 생성 (픽셀보다 1칸 더 필요)
pixel_edges = np.arange(-0.5, 28, 1)

plt.figure(figsize=(8, 8))
plt.imshow(sample_img, cmap='gray')

# 픽셀 경계에 격자 그리기
plt.grid(color='red', linewidth=0.5)
plt.xticks(pixel_edges)
plt.yticks(pixel_edges)

plt.tick_params(bottom=False, left=False, labelbottom=False, labelleft=False)
plt.title(f"MNIST Image with Pixel Grid (Label: {sample_label})")
plt.show()
```
