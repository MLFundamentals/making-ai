```python
!pip -q install gTTS librosa numpy

from gtts import gTTS
from IPython.display import Audio, display
import librosa

text = "속보입니다. 서울시는 오늘 오전 9시를 기해 인공지능 기반 교통 신호 제어 시스템을 공식적으로 가동했다고 발표했습니다. 자세한 소식 전해 드립니다."
gTTS(text=text, lang="ko", slow=False).save("ko_tts.mp3")

# 재생 속도 조절을 위해 mp3를 librosa로 로드
y, sr = librosa.load("ko_tts.mp3", sr=None)

# 1.3배속 적용
y_fast = librosa.effects.time_stretch(y, rate=1.3)

print("▶ 재생 (1.3x speed)")
display(Audio(y_fast, rate=sr))
```
