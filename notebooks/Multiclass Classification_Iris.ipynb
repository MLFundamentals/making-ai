```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import OneHotEncoder

# Iris 데이터셋 불러오기
iris = load_iris()
X = iris.data  # (150, 4)
y = iris.target.reshape(-1, 1)  # (150, 1)

# 원-핫 인코딩
encoder = OneHotEncoder(sparse_output=False)
y_encoded = encoder.fit_transform(y)  # (150, 3)

# 소프트맥스 함수 정의
def softmax(z):
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))  # overflow 방지
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)

# 가중치 초기화
X = np.hstack([X, np.ones((X.shape[0], 1))])  # Bias Trick
n_classes = y_encoded.shape[1]
W = np.random.randn(X.shape[1], n_classes) * 0.01

# 하이퍼파라미터 설정
learning_rate = 0.1
epochs = 1000

# 학습 진행
for epoch in range(epochs + 1):
    logits = np.dot(X, W)
    y_pred = softmax(logits)

    # 크로스 엔트로피 손실 계산
    loss = -np.mean(np.sum(y_encoded * np.log(y_pred + 1e-15), axis=1))

    # 그래디언트 계산
    gradient = X.T.dot(y_pred - y_encoded) / X.shape[0]

    # 가중치 업데이트
    W -= learning_rate * gradient

    # 중간 cost 출력
    if epoch % 100 == 0:
        print(f'Epoch {epoch}, Cost: {loss:.4f}')

# 정확도 계산 - 학습 데이터 전체를 테스트 데이터로 사용
pred_labels = np.argmax(y_pred, axis=1)
accuracy = np.mean(pred_labels == y.flatten())
print(f'\n모델 정확도: {accuracy:.4f}')

#======================== 아래 코드는 책에서는 생략 ============================

# 사용자 입력 테스트
print("\n\n▶ 사용자 입력 테스트:")
print("꽃받침 길이, 꽃받침 너비, 꽃잎 길이, 꽃잎 너비(cm)를 순서대로 입력하세요.\n(예: 5.1, 3.5, 1.4, 0.2)\n")
features = [float(input(f'특징 {i+1}: ')) for i in range(4)]
x_input = np.array(features + [1.0]).reshape(1, -1)  # Bias Trick

user_pred = softmax(np.dot(x_input, W))
user_pred_class = np.argmax(user_pred)
emojis = {0: '🌸 Setosa', 1: '🌼 Versicolor', 2: '🌺 Virginica'}
print(f'\n➜ 예측된 붓꽃 품종: {emojis.get(user_pred_class, "❓ Unknown")}')
```
