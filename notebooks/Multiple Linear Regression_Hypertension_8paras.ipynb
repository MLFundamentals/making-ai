```python
import numpy as np
import pandas as pd
import gspread
from google.auth import default
from google.colab import auth
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDRegressor

# Google Sheets 인증 및 데이터 불러오기
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# Google Sheets 파일 열기
spreadsheet_id = "1eC4YnMRr5a6v2EmxcZBUb-mGv6k8Qz-8UcNTSoeM2qU"
spreadsheet = gc.open_by_key(spreadsheet_id)
worksheet = spreadsheet.sheet1  # 첫번째 시트 선택
data = worksheet.get_all_values()

# DataFrame 변환 및 헤더 지정
df = pd.DataFrame(data[1:], columns=data[0])

# 데이터 타입 변환
df['Age'] = df['Age'].astype(int)
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
df['BMI'] = df['BMI'].astype(float)
df['Diet'] = df['Diet'].map({'Poor': 0, 'Average': 1, 'Good': 2})
df['Exercise_per_week'] = df['Exercise_per_week'].astype(int)
df['Alcohol_intake'] = df['Alcohol_intake'].astype(float)
df['Smoking'] = df['Smoking'].astype(int)
df['Family_history'] = df['Family_history'].astype(int)
df['Hypertension_risk'] = df['Hypertension_risk'].astype(float)

# 특성값 데이터(X)와 레이블 데이터(y) 분리
X = df.drop(columns=['Hypertension_risk'])
y = df['Hypertension_risk']

# 데이터 정규화(표준화)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# SGD를 활용한 다중 선형회귀 모델 학습 (Epoch별 학습과정 출력)
sgd_model = SGDRegressor(max_iter=1000, eta0=0.01, learning_rate="adaptive", verbose=1)
sgd_model.fit(X_scaled, y)

# 새 환자 데이터 입력 및 예측
print(f"\n\n▶ 새 환자의 고혈압 발병 확률 예측하기\n")

patient_info = [[
    int(input("나이: ")),
    int(input("성별 (남성:0, 여성:1): ")),
    float(input("BMI (체중을 키의 제곱으로 나눈 값. 보통 20~25): ")),
    int(input("식습관 (좋음:2, 보통:1, 나쁨:0): ")),
    int(input("주간 운동 횟수: ")),
    float(input("음주량 (주당 평균 잔): ")),
    int(input("흡연량 (일일 개비 수): ")),
    int(input("고혈압 가족력 (있음:1, 없음:0): "))
]]

new_patient_df = pd.DataFrame(patient_info, columns=X.columns)
new_patient_scaled = scaler.transform(new_patient_df)
predicted_risk = sgd_model.predict(new_patient_scaled)
print(f"\n예측된 고혈압 발병 확률: {predicted_risk[0]*100:.2f}%")


#-------------------------------------------------------------------------------#
# 학습 완료 후 회귀모델 수식 출력해보기
#print("\n▶ 학습이 완료된 다중 선형회귀 모델 수식")
#feature_names = [f"x{i+1}" for i in range(len(X.columns))]  # x1, x2, ... 형식으로 변수 이름 변경
#coefficients = sgd_model.coef_
#intercept = sgd_model.intercept_
#
#equation = "H(x) = " + " + ".join([f"{coeff:.4f} * {feature}" for coeff, feature in zip(coefficients, feature_names)])
#equation += f" + {intercept[0]:.4f}"
#print(equation)

#-------------------------------------------------------------------------------#
# 구글시트에서 데이터 바로 불러오기
#google_sheet_id = "1eC4YnMRr5a6v2EmxcZBUb-mGv6k8Qz-8UcNTSoeM2qU"
#sheet_name = "Sheet1"
#url = f"https://docs.google.com/spreadsheets/d/{google_sheet_id}/gviz/tq?tqx=out:csv&sheet={sheet_name}"
#df = pd.read_csv(url)
```
