```python
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

(X_train, y_train), _ = tf.keras.datasets.mnist.load_data()

sample_img = X_train[20000]
sample_label = y_train[20000]

plt.figure(figsize=(12, 10))
sns.heatmap(sample_img, annot=True, cmap='gray', fmt='d', annot_kws={"size":6}, cbar=False)
plt.title("MNIST Image as Numeric Pixel Values (0~255)")
plt.xlabel("Pixels (X)")
plt.ylabel("Pixels (Y)")
plt.show()
```
