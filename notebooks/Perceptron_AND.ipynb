```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD

# 학습 데이터 (AND 연산)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([[0], [0], [0], [1]])

# 모델 구성
model = Sequential([
    Input(shape=(2,)),
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy',
              optimizer=SGD(learning_rate=0.5),
              metrics=['accuracy'])

# 학습
model.fit(X, y, epochs=200, verbose=0)

# 학습된 파라미터 확인
weights, bias = model.layers[0].get_weights()
w1, w2 = weights[0][0], weights[1][0]
b = bias[0]

print("\n[학습 완료된 파라미터]")
print(f"가중치(w1, w2): [{w1:.4f}, {w2:.4f}]")
print(f"바이어스(b): {b:.4f}")

# 테스트
print("\n[학습 완료된 AND 연산자 모델 테스트]")
for x in X:
    prob = model.predict(np.array([x]), verbose=0)[0][0]
    pred = int(prob >= 0.5)
    print(f"입력: {x} → 예측: {pred} (확률: {prob:.4f})")

# 사용자 입력 반복
while True:
    user_input = input("\n두 개의 이진 입력값을 공백으로 구분하여 입력하세요 (예: '1 0') 또는 'q'를 입력하여 종료: ")
    if user_input.strip().lower() == 'q':
        break
    try:
        parts = user_input.strip().split()
        if len(parts) != 2 or not all(p in ['0', '1'] for p in parts):
            raise ValueError
        x_input = np.array([[int(parts[0]), int(parts[1])]])
        prob = model.predict(x_input, verbose=0)[0][0]
        pred = int(prob >= 0.5)
        print(f"모델 예측 결과: {pred} (확률: {prob:.4f})")
    except:
        print("입력 형식이 잘못되었습니다. 예: '1 0'")
```
