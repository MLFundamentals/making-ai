```python
# ============================================
# 🧩 마스킹 단어 예측기 체험 (Gradio)
# ============================================

!pip -q install transformers gradio

from transformers import pipeline
import gradio as gr

# fill-mask 파이프라인
unmasker = pipeline("fill-mask", model="bert-base-uncased")

# 예측 함수
def predict_masked(text):
    predictions = unmasker(text)
    results = [f"{i+1}. {p['sequence']} (score: {p['score']:.2%})" for i, p in enumerate(predictions)]
    return "\n".join(results)

# Gradio UI
gr.Interface(
    fn=predict_masked,
    inputs=gr.Textbox(lines=2, label="📝 [MASK]가 포함된 문장", value="The capital of [MASK] is Seoul."),
    outputs=gr.Textbox(label="🔍 예측 결과 (Top 5)"),
    title="🧠 마스킹 단어 예측기 (BERT)",
    description="문장에서 [MASK] 자리에 들어갈 단어를 AI가 예측합니다. 영어 문장만 지원합니다.",
).launch()
```

```python
# ============================================
# 🇰🇷 한국어 문장 유사도 체험 (Gradio)
# ============================================

!pip -q install sentence-transformers gradio

from sentence_transformers import SentenceTransformer, util
import gradio as gr

# 한국어 SBERT 모델 로드
model = SentenceTransformer("jhgan/ko-sbert-nli")

# 유사도 계산 함수
def compare_ko_similarity(sent1, sent2):
    emb1 = model.encode(sent1, convert_to_tensor=True)
    emb2 = model.encode(sent2, convert_to_tensor=True)
    score = util.cos_sim(emb1, emb2).item()
    return f"🔁 유사도 점수: {score:.2%}"

# Gradio UI
gr.Interface(
    fn=compare_ko_similarity,
    inputs=[
        gr.Textbox(lines=2, label="문장 1", value="나는 오늘 아침에 커피를 마셨어."),
        gr.Textbox(lines=2, label="문장 2", value="오늘 아침에 커피를 마신 기억이 난다."),
    ],
    outputs=gr.Textbox(label="📊 유사도 결과"),
    title="🧠 한국어 문장 유사도 측정기 (KoSBERT)",
    description="두 한국어 문장의 의미 유사도를 AI가 판단해 점수로 보여줍니다.",
).launch()
```

```python
# ============================================
# 😊 감정 분석기 체험 (영어 / Gradio)
# ============================================

!pip -q install transformers gradio

from transformers import pipeline
import gradio as gr

# 감정 분석 파이프라인
classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")

# 분석 함수
def analyze_sentiment(text):
    results = classifier(text)
    label = results[0]['label']
    score = results[0]['score']
    emoji = "😊" if label == "POSITIVE" else "😞"
    return f"{emoji} {label} ({score:.2%} 확신)"

# Gradio 인터페이스
gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(lines=2, label="📝 문장을 입력하세요", value="I really enjoyed this movie!"),
    outputs=gr.Textbox(label="📊 감정 분석 결과"),
    title="🧠 영어 감정 분석기",
    description="영어 문장을 입력하면, AI가 긍정 또는 부정 감정을 분석해줍니다.",
).launch()
```
