```python
# ============================================
# 🇰🇷 KoGPT2 텍스트 생성 체험기 (Gradio)
# ============================================

!pip -q install transformers gradio

from transformers import GPT2LMHeadModel, PreTrainedTokenizerFast, pipeline
import gradio as gr

# KoGPT2 모델 및 토크나이저 불러오기
model_name = "skt/kogpt2-base-v2"
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = PreTrainedTokenizerFast.from_pretrained(
    model_name,
    bos_token='</s>', eos_token='</s>', unk_token='<unk>',
    pad_token='<pad>', mask_token='<mask>'
)

# 텍스트 생성 파이프라인
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)

# 생성 함수
def generate_text(prompt, max_len, num_return):
    outputs = generator(prompt, max_new_tokens=max_len, num_return_sequences=num_return, do_sample=True)
    return '\n\n'.join([f"[{i+1}]\n" + o['generated_text'] for i, o in enumerate(outputs)])

# Gradio 인터페이스
gr.Interface(
    fn=generate_text,
    inputs=[
        gr.Textbox(lines=2, label="📝 시작 문장", value="요즘 가장 관심 있는 주제는"),
        gr.Slider(20, 50, step=5, value=30, label="📏 최대 생성 길이"),
        gr.Slider(1, 3, step=1, value=1, label="🔁 생성 문장 수"),
    ],
    outputs=gr.Textbox(label="✍️ 생성된 문장들"),
    title="🇰🇷 한국어 GPT-2 텍스트 생성기",
    description="사용자가 입력한 문장을 바탕으로 한국어 문장을 생성합니다. 다양한 문장을 입력하며 실험해보세요!",
).launch()
```
