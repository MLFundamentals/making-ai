```python
# 필요한 라이브러리 설치
!pip install -q transformers sentencepiece

# Hugging Face Transformers 불러오기
from transformers import MBartForConditionalGeneration, MBart50TokenizerFast
import torch

# 사전학습된 mBART-50 모델과 토크나이저 로드
model_name = "facebook/mbart-large-50-many-to-many-mmt"
tokenizer = MBart50TokenizerFast.from_pretrained(model_name)
model = MBartForConditionalGeneration.from_pretrained(model_name)

# 원본 언어(한국어)와 목표 언어(영어) 설정
tokenizer.src_lang = "ko_KR"       # 입력 언어: 한국어
target_lang_token = "en_XX"        # 출력 언어: 영어

# 번역 함수 정의
def translate(text):
    # 입력 문장을 토큰화 → 텐서 변환
    inputs = tokenizer(text, return_tensors="pt")
    # 번역 결과 토큰 생성 (출력 언어 지정)
    generated_tokens = model.generate(
        **inputs,
        forced_bos_token_id=tokenizer.lang_code_to_id[target_lang_token]
    )
    # 토큰을 사람이 읽을 수 있는 문자열로 변환
    translated = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]
    return translated

# 사용자 입력 반복 실행
while True:
    sentence = input("\n💬 한국어 문장을 입력하세요 (종료하려면 'quit' 입력): ")
    if sentence.strip().lower() == 'quit':   # 'quit' 입력 시 종료
        break
    result = translate(sentence)             # 번역 실행
    print("👉 번역 결과:", result)           # 결과 출력
```
